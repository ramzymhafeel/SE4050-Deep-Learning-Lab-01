## Task 1.1 — Random 4×4 Array from an Exponential Distribution

### Introduction

The exponential distribution is a continuous probability distribution commonly used to model the time between independent events.

In this task, NumPy is used to generate a random 4×4 array from an exponential distribution.

### Exponential Distribution Parameters

NumPy generates exponentially distributed random values using:

`np.random.exponential(scale, size)`

Where:

- `scale` is the inverse of the rate parameter
- `size` defines the output dimensions
- `size=(4, 4)` creates four rows and four columns

For this task, the scale value is set to `1.0`.

```python
import numpy as np

# Set a fixed random seed for reproducible results.
np.random.seed(42)

print("Random seed set successfully.")
```

```python
# Generate a random 4×4 array from an exponential distribution.

exponential_array = np.random.exponential(
    scale=1.0,
    size=(4, 4)
)

print("Random 4×4 array from an exponential distribution:")
print(exponential_array)
```

```python
# Display values rounded to four decimal places.

print("Exponential array rounded to four decimal places:\n")
print(np.round(exponential_array, 4))
```

```python
print("Array shape:", exponential_array.shape)
print("Number of dimensions:", exponential_array.ndim)
print("Total number of elements:", exponential_array.size)
```

```python
print("Array data type:", exponential_array.dtype)
```

```python
array_mean = np.mean(exponential_array)
array_std = np.std(exponential_array)
array_min = np.min(exponential_array)
array_max = np.max(exponential_array)

print("Basic statistics of the generated array:")
print("Mean:", round(array_mean, 4))
print("Standard deviation:", round(array_std, 4))
print("Minimum value:", round(array_min, 4))
print("Maximum value:", round(array_max, 4))
```

```python
all_non_negative = np.all(exponential_array >= 0)

print("Are all values non-negative?", all_non_negative)
```

```python
import pandas as pd

array_df = pd.DataFrame(
    exponential_array,
    columns=["Column 1", "Column 2", "Column 3", "Column 4"],
    index=["Row 1", "Row 2", "Row 3", "Row 4"]
)

array_df.round(4)
```

### Interpretation

A 4×4 NumPy array containing 16 random values was generated successfully from an exponential distribution.

The array contains only non-negative floating-point values, which is consistent with the properties of an exponential distribution.

The shape `(4, 4)` confirms that the generated array contains four rows and four columns.

```python
# Task 1.1: Generate a random 4×4 array
# from an exponential distribution.

import numpy as np

np.random.seed(42)

exponential_array = np.random.exponential(
    scale=1.0,
    size=(4, 4)
)

print("Random 4×4 exponential-distribution array:\n")
print(np.round(exponential_array, 4))

print("\nArray shape:", exponential_array.shape)
print("Number of elements:", exponential_array.size)
print("All values are non-negative:",
      np.all(exponential_array >= 0))
```

## Task 1.2 — Visualization of Exponential, Uniform, and Normal Distributions

### Introduction

In this task, three probability distributions are generated and compared:

- Exponential distribution
- Uniform distribution
- Normal distribution

Each distribution contains 100,000 random values. Their probability density shapes are visualized using histograms.

### Distribution Characteristics

**Exponential distribution:**  
Produces non-negative values and is usually right-skewed.

**Uniform distribution:**  
Produces values with approximately equal probability across a specified range.

**Normal distribution:**  
Produces a symmetric bell-shaped distribution centered around the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the random seed for reproducible results.
np.random.seed(42)

print("Random seed set successfully.")
```

```python
# Generate 100,000 values from an exponential distribution.
# The reshape operation creates a 100000 × 1 array.

exponential_data = np.random.exponential(
    scale=1.0,
    size=100000
).reshape(100000, 1)

print("Exponential data shape:", exponential_data.shape)
print("First five values:\n", exponential_data[:5])
```

```python
# Generate 100,000 values from a uniform distribution
# between 0 and 1.

uniform_data = np.random.uniform(
    low=0.0,
    high=1.0,
    size=100000
)

print("Uniform data shape:", uniform_data.shape)
print("First five values:", uniform_data[:5])
```

```python
# Generate 100,000 values from a standard normal distribution.
# Mean = 0 and standard deviation = 1.

normal_data = np.random.normal(
    loc=0.0,
    scale=1.0,
    size=100000
)

print("Normal data shape:", normal_data.shape)
print("First five values:", normal_data[:5])
```

```python
exponential_plot_data = exponential_data.flatten()

print("Original shape:", exponential_data.shape)
print("Plotting shape:", exponential_plot_data.shape)
```

```python
plt.figure(figsize=(12, 7))

plt.hist(
    exponential_plot_data,
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Exponential Distribution"
)

plt.hist(
    uniform_data,
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Uniform Distribution"
)

plt.hist(
    normal_data,
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Normal Distribution"
)

plt.axis([-3, 6, 0, 1.2])

plt.title(
    "Comparison of Exponential, Uniform, and Normal Distributions",
    fontsize=14
)

plt.xlabel("Value", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)

plt.legend(loc="upper right")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

### Selection of Histogram Bins

The original example uses 100 bins. In this implementation, the number of bins was increased to 150.

Using 150 bins provides a more detailed view because the dataset contains 100,000 observations. It improves the visibility of each distribution while keeping the graph readable.

### Plot View

The x-axis is limited to the range `[-3, 6]` so that the important regions of all three distributions can be compared clearly.

The y-axis is limited to `[0, 1.2]` to keep the highest-density regions visible without creating excessive empty space.

```python
print("Exponential distribution:")
print("Mean:", round(np.mean(exponential_plot_data), 4))
print("Standard deviation:",
      round(np.std(exponential_plot_data), 4))
print("Minimum:", round(np.min(exponential_plot_data), 4))
print("Maximum:", round(np.max(exponential_plot_data), 4))

print("\nUniform distribution:")
print("Mean:", round(np.mean(uniform_data), 4))
print("Standard deviation:",
      round(np.std(uniform_data), 4))
print("Minimum:", round(np.min(uniform_data), 4))
print("Maximum:", round(np.max(uniform_data), 4))

print("\nNormal distribution:")
print("Mean:", round(np.mean(normal_data), 4))
print("Standard deviation:",
      round(np.std(normal_data), 4))
print("Minimum:", round(np.min(normal_data), 4))
print("Maximum:", round(np.max(normal_data), 4))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].hist(
    exponential_plot_data,
    bins=100,
    density=True,
    edgecolor="black"
)
axes[0].set_title("Exponential Distribution")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Density")
axes[0].set_xlim(0, 6)
axes[0].grid(alpha=0.3)

axes[1].hist(
    uniform_data,
    bins=100,
    density=True,
    edgecolor="black"
)
axes[1].set_title("Uniform Distribution")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Density")
axes[1].set_xlim(0, 1)
axes[1].grid(alpha=0.3)

axes[2].hist(
    normal_data,
    bins=100,
    density=True,
    edgecolor="black"
)
axes[2].set_title("Normal Distribution")
axes[2].set_xlabel("Value")
axes[2].set_ylabel("Density")
axes[2].set_xlim(-4, 4)
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Interpretation

The three histograms demonstrate clearly different probability-distribution shapes.

- The exponential distribution is positively skewed. Most values occur near zero, while fewer large values appear toward the right.
- The uniform distribution is approximately flat between zero and one, showing that values in this interval have similar probabilities.
- The normal distribution is approximately symmetric and bell-shaped around zero.

The visualization confirms that probability distributions can have different ranges, levels of symmetry, and density patterns.

```python
# Task 1.2: Compare exponential, uniform,
# and normal distributions.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

exponential_data = np.random.exponential(
    scale=1.0,
    size=100000
).reshape(100000, 1)

uniform_data = np.random.uniform(
    low=0.0,
    high=1.0,
    size=100000
)

normal_data = np.random.normal(
    loc=0.0,
    scale=1.0,
    size=100000
)

plt.figure(figsize=(12, 7))

plt.hist(
    exponential_data.flatten(),
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Exponential"
)

plt.hist(
    uniform_data,
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Uniform"
)

plt.hist(
    normal_data,
    density=True,
    bins=150,
    histtype="step",
    linewidth=2,
    label="Normal"
)

plt.axis([-3, 6, 0, 1.2])
plt.title("Random Distribution Comparison")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend(loc="upper right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("Exponential array shape:", exponential_data.shape)
```
