## Task 2.2 — Select the Required Features

### Feature Selection

A new dataframe is created using only the four features required by the lab sheet:

- `N_total`
- `N_bulk`
- `N_surface`
- `R_avg`

The original dataframe remains unchanged.

```python
new_df = au_df[
    [
        "N_total",
        "N_bulk",
        "N_surface",
        "R_avg"
    ]
].copy()

print("Filtered dataframe created successfully.")
print("Filtered dataframe shape:", new_df.shape)
```

```python
print("Filtered dataframe columns:")
print(new_df.columns.tolist())
```

```python
print("Data types of selected features:\n")
print(new_df.dtypes)
```

```python
for column in required_columns:
    new_df[column] = pd.to_numeric(
        new_df[column],
        errors="coerce"
    )

print("Selected features converted to numeric format.")
```

```python
print("Missing values in selected features:\n")
print(new_df.isnull().sum())
```

### Interpretation

The Au nanoparticle dataset was loaded successfully using Pandas.

A new dataframe named `new_df` was created containing only the four required features: `N_total`, `N_bulk`, `N_surface`, and `R_avg`.

The filtered dataframe contains four columns, and its first 20 samples were displayed successfully. The original dataset remains unchanged.

```python
# Task 2.1 to Task 2.3:
# Load the Au nanoparticle dataset,
# select the required features,
# and display the first 20 samples.

import os
import pandas as pd

dataset_folder = (
    "/content/drive/MyDrive/"
    "SE4050_DeepLearning/"
    "Lab_01/"
    "02_Datasets"
)

au_file_path = os.path.join(
    dataset_folder,
    "Au_nanoparticle_dataset.csv"
)

au_df = pd.read_csv(au_file_path)
au_df.columns = au_df.columns.str.strip()

selected_features = [
    "N_total",
    "N_bulk",
    "N_surface",
    "R_avg"
]

new_df = au_df[selected_features].copy()

for column in selected_features:
    new_df[column] = pd.to_numeric(
        new_df[column],
        errors="coerce"
    )

print("Original dataset shape:", au_df.shape)
print("Filtered dataframe shape:", new_df.shape)
print("\nFirst 20 samples:")

display(new_df.head(20))
```

```python
assert new_df.shape[1] == 4, (
    "The filtered dataframe must contain exactly four columns."
)

assert list(new_df.columns) == required_columns, (
    "The selected columns are incorrect or in the wrong order."
)

assert len(new_df.head(20)) == 20, (
    "The output must contain exactly 20 samples."
)

print("All Phase 6 validation tests passed.")
```
