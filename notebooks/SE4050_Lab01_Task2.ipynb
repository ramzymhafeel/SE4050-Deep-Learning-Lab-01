```python
coefficient_of_variation = (
    statistics_df.std()
    / statistics_df.mean()
)

print("Coefficient of variation:\n")
print(coefficient_of_variation.round(4))
```

### Interpretation

The descriptive statistics summarize the central tendency and variability of the four selected Au nanoparticle features.

- The **mean** represents the average value of each feature.
- The **standard deviation** indicates how widely values are distributed around the mean.
- The **first quartile (`Q1`)** represents the value below which approximately 25% of observations fall.
- The **median (`Q2`)** divides the data into two equal halves.
- The **third quartile (`Q3`)** represents the value below which approximately 75% of observations fall.
- The **interquartile range (`IQR`)** represents the spread of the middle 50% of observations.

Differences between the mean and median may indicate that a feature has a skewed distribution. A large standard deviation or IQR indicates that the values of that feature are more widely dispersed.

```python
for column in statistics_df.columns:
    mean_value = statistics_df[column].mean()
    median_value = statistics_df[column].median()

    if mean_value > median_value:
        shape_message = "may be positively skewed"
    elif mean_value < median_value:
        shape_message = "may be negatively skewed"
    else:
        shape_message = "appears approximately symmetric"

    print(
        f"{column}: mean = {mean_value:.4f}, "
        f"median = {median_value:.4f}; "
        f"the distribution {shape_message}."
    )
```

### Dataset-Specific Observation

The mean and median values are not identical for all four features. This suggests that some of the distributions may be skewed.

The histogram analysis in the next section will be used to examine the shapes of these distributions visually.

```python
# Task 2.4:
# Calculate mean, standard deviation, and quartile values.

statistics_df = new_df.dropna().copy()

statistics_table = pd.DataFrame({
    "Mean": statistics_df.mean(),
    "Standard Deviation": statistics_df.std(),
    "Q1 (25%)": statistics_df.quantile(0.25),
    "Median (50%)": statistics_df.quantile(0.50),
    "Q3 (75%)": statistics_df.quantile(0.75)
})

statistics_table["IQR"] = (
    statistics_table["Q3 (75%)"]
    - statistics_table["Q1 (25%)"]
)

statistics_table.index.name = "Feature"

print("Mean, standard deviation, and quartile values:")

display(statistics_table.round(4))
```
