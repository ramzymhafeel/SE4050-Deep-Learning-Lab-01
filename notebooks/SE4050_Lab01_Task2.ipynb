### Interpretation

The descriptive statistics summarize the central tendency and variability of the four selected Au nanoparticle features.

- The **mean** represents the average value of each feature.
- The **standard deviation** indicates how widely values are distributed around the mean.
- The **first quartile (`Q1`)** represents the value below which approximately 25% of observations fall.
- The **median (`Q2`)** divides the data into two equal halves.
- The **third quartile (`Q3`)** represents the value below which approximately 75% of observations fall.
- The **interquartile range (`IQR`)** represents the spread of the middle 50% of observations.

Differences between the mean and median may indicate that a feature has a skewed distribution. A large standard deviation or IQR indicates that the values of that feature are more widely dispersed.

```python
for column in statistics_df.columns:
    mean_value = statistics_df[column].mean()
    median_value = statistics_df[column].median()

    if mean_value > median_value:
        shape_message = "may be positively skewed"
    elif mean_value < median_value:
        shape_message = "may be negatively skewed"
    else:
        shape_message = "appears approximately symmetric"

    print(
        f"{column}: mean = {mean_value:.4f}, "
        f"median = {median_value:.4f}; "
        f"the distribution {shape_message}."
    )
```

## Task 2.5 — Histograms in a 1×4 Layout

### Histogram Analysis

A histogram displays the frequency distribution of a numerical feature by dividing its values into intervals called bins.

The following four Au nanoparticle features are visualized in a 1×4 layout:

- `N_total`
- `N_bulk`
- `N_surface`
- `R_avg`

The histograms help identify the distribution shape, spread, concentration of values, and possible skewness.

```python
print("Dataframe shape:", new_df.shape)
print("Selected features:", new_df.columns.tolist())

display(new_df.head())
```

```python
plot_df = new_df.dropna().copy()

print("Original number of rows:", new_df.shape[0])
print("Rows available for plotting:", plot_df.shape[0])
```

```python
import matplotlib.pyplot as plt

selected_features = [
    "N_total",
    "N_bulk",
    "N_surface",
    "R_avg"
]

fig, axes = plt.subplots(
    1,
    4,
    figsize=(20, 5)
)

for axis, feature in zip(axes, selected_features):
    axis.hist(
        plot_df[feature],
        bins=30,
        edgecolor="black",
        alpha=0.8
    )

    axis.set_title(
        f"Distribution of {feature}",
        fontsize=12
    )

    axis.set_xlabel(feature)
    axis.set_ylabel("Frequency")
    axis.grid(
        axis="y",
        alpha=0.3
    )

fig.suptitle(
    "Histograms of Selected Au Nanoparticle Features",
    fontsize=16,
    y=1.04
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
```

```python
fig, axes = plt.subplots(
    1,
    4,
    figsize=(22, 5)
)

for axis, feature in zip(axes, selected_features):
    axis.hist(
        plot_df[feature],
        bins="auto",
        edgecolor="black",
        alpha=0.8
    )

    axis.axvline(
        plot_df[feature].mean(),
        linestyle="--",
        linewidth=2,
        label="Mean"
    )

    axis.axvline(
        plot_df[feature].median(),
        linestyle=":",
        linewidth=2,
        label="Median"
    )

    axis.set_title(feature, fontsize=13)
    axis.set_xlabel("Value")
    axis.set_ylabel("Frequency")
    axis.grid(axis="y", alpha=0.3)
    axis.legend()

fig.suptitle(
    "Distribution of Au Nanoparticle Features",
    fontsize=16,
    y=1.05
)

plt.tight_layout()
plt.show()
```

### Mean and Median Indicators

The dashed line represents the mean, while the dotted line represents the median.

When the mean and median are close together, the distribution may be relatively symmetric. When they are separated, the distribution may be skewed.

The histogram shape should also be inspected before making a final conclusion.

```python
print("Distribution shape indicators:\n")

for feature in selected_features:
    mean_value = plot_df[feature].mean()
    median_value = plot_df[feature].median()
    skewness_value = plot_df[feature].skew()

    print(feature)
    print("  Mean:", round(mean_value, 4))
    print("  Median:", round(median_value, 4))
    print("  Skewness:", round(skewness_value, 4))
    print()
```

### Histogram Interpretation

The four histograms show that the selected Au nanoparticle features have different ranges and distribution patterns.

Some features may contain a concentration of observations at lower values with a longer tail toward higher values. This indicates possible positive skewness.

The distance between the mean and median lines also provides an initial indication of distribution asymmetry. However, the histogram shapes and calculated skewness values should be considered together.

## Task 2.6 — Seaborn Pairplot

### Seaborn Pairplot

A Seaborn pairplot displays pairwise relationships among numerical features.

- The diagonal plots show the distribution of each individual feature.
- The off-diagonal plots show scatter plots between pairs of features.

This visualization helps identify relationships, trends, clusters, unusual observations, and possible correlations.

```python
import seaborn as sns
import matplotlib.pyplot as plt

pair_plot = sns.pairplot(
    plot_df[
        [
            "N_total",
            "N_bulk",
            "N_surface",
            "R_avg"
        ]
    ],
    diag_kind="hist",
    plot_kws={
        "alpha": 0.6,
        "s": 25,
        "edgecolor": "none"
    },
    diag_kws={
        "bins": 30,
        "edgecolor": "black"
    }
)

pair_plot.fig.suptitle(
    "Pairplot of Au Nanoparticle Features",
    y=1.02,
    fontsize=16
)

plt.show()
```

```python
pair_plot = sns.pairplot(
    plot_df[selected_features],
    diag_kind="hist",
    corner=False,
    height=2.7,
    aspect=1,
    plot_kws={
        "alpha": 0.5,
        "s": 20
    },
    diag_kws={
        "bins": 25
    }
)

pair_plot.fig.suptitle(
    "Scatter Plots and Histograms of Selected Features",
    y=1.02,
    fontsize=15
)

plt.show()
```

```python
corner_pairplot = sns.pairplot(
    plot_df[selected_features],
    diag_kind="hist",
    corner=True,
    height=2.8,
    plot_kws={
        "alpha": 0.6,
        "s": 25
    },
    diag_kws={
        "bins": 30
    }
)

corner_pairplot.fig.suptitle(
    "Lower-Triangle Pairplot of Au Nanoparticle Features",
    y=1.02,
    fontsize=15
)

plt.show()
```

```python
pairplot_correlation = (
    plot_df[selected_features]
    .corr()
)

print("Pearson correlation matrix:")

display(pairplot_correlation.round(3))
```

```python
import numpy as np

correlation_without_diagonal = (
    pairplot_correlation
    .where(
        ~np.eye(
            pairplot_correlation.shape[0],
            dtype=bool
        )
    )
)

strongest_pair = (
    correlation_without_diagonal
    .abs()
    .stack()
    .idxmax()
)

strongest_value = pairplot_correlation.loc[
    strongest_pair[0],
    strongest_pair[1]
]

print("Strongest feature relationship:")
print(
    strongest_pair[0],
    "and",
    strongest_pair[1]
)

print(
    "Correlation coefficient:",
    round(strongest_value, 4)
)
```

### Pairplot Interpretation

The diagonal histograms show the individual distributions of `N_total`, `N_bulk`, `N_surface`, and `R_avg`.

The off-diagonal scatter plots show how each pair of features changes together. A narrow upward pattern indicates a positive relationship, while a downward pattern indicates a negative relationship. A widely scattered pattern indicates a weaker relationship.

The pairplot can also reveal nonlinear patterns, clusters, and possible outliers that may not be visible from descriptive statistics alone.

```python
# Task 2.5:
# Plot the four selected feature histograms
# in the required 1×4 layout.

selected_features = [
    "N_total",
    "N_bulk",
    "N_surface",
    "R_avg"
]

plot_df = new_df[selected_features].dropna().copy()

fig, axes = plt.subplots(
    1,
    4,
    figsize=(20, 5)
)

for axis, feature in zip(axes, selected_features):
    axis.hist(
        plot_df[feature],
        bins=30,
        edgecolor="black",
        alpha=0.8
    )

    axis.set_title(feature)
    axis.set_xlabel("Value")
    axis.set_ylabel("Frequency")
    axis.grid(axis="y", alpha=0.3)

fig.suptitle(
    "Histograms of Au Nanoparticle Features",
    fontsize=16,
    y=1.04
)

plt.tight_layout()
plt.show()
```
